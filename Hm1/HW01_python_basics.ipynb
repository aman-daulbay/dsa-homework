## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.

9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
def sum_city(city):
    sum_arr=0
    for i in range(len(cities_clean)):
        if cities_clean[i]==city and paid[i]:
            sum_arr+=amounts_int[i]

    return sum_arr

print("Almaty:",sum_city("Almaty"))
print("Astana:",sum_city("Astana"))
````

````python
index=0
max_amount=amounts_int[0]
for i in range(len(amounts_int)):
    if amounts_int[i]>max_amount and paid[i]:
        max_amount=amounts_int[i]
        index=i

print(f"Index: {index}, City: {cities_clean[index]}")
````

````python
count_weekend=list(filter(lambda i: i=="сб" or i=="вс",days))
print(len(count_weekend)/len(days)*100,"%")
````

13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# Челлендж
unique_cities=[]
for city in cities_clean:
    if city not in unique_cities:
        unique_cities.append(city)
print(unique_cities)
revenue_by_city=[0]*len(unique_cities)
for i in range(len(amounts_int)):
    if paid[i]:
        revenue_by_city[unique_cities.index(cities_clean[i])]+=amounts_int[i]

print(revenue_by_city)
````

````python
pairs=list(zip(unique_cities,revenue_by_city))
pairs=sorted(pairs, key=lambda p:p[1], reverse=True)

unique_cities=[p[0] for p in pairs]
revenue_by_city=[p[1] for p in pairs]

print(unique_cities)
print(revenue_by_city)
````
